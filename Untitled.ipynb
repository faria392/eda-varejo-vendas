INICIALIZANDO TRATAMENTO DOS DADOS:

```python
dados = dados.drop_duplicates()
```

```python
print("Linhas duplicadas antes:", dados.duplicated().sum())

duplicadas = dados[dados.duplicated(keep=False)]
print(duplicadas.shape)

dados = dados.drop_duplicates(keep='first').reset_index(drop=True)

print("Linhas duplicadas depois:", dados.duplicated().sum())
print("Shape final:", dados.shape)
```

```python
print(dados.shape)
print(dados.duplicated().sum())
```

```python
mapa_gordura = {
    'baixo teor de gordura': 'Baixo Teor de Gordura',
    'BTG': 'Baixo Teor de Gordura',
    'Baixo Teor de Gordura': 'Baixo Teor de Gordura',
    'reg': 'Regular',
    'Regular': 'Regular'
}

dados['item_conteudo_gordura'] = dados['item_conteudo_gordura'].str.strip().replace(mapa_gordura)

print(dados['item_conteudo_gordura'].value_counts(dropna=False))
```

```python
pesos_por_item = dados.groupby('item_identificador')['item_peso'].nunique()
print(pesos_por_item.value_counts())

itens_sem_peso_algum = dados.groupby('item_identificador')['item_peso'].apply(lambda x: x.isna().all())
print("Itens sem nenhum peso registrado:", itens_sem_peso_algum.sum())
```

```python
dados['item_peso'] = dados.groupby('item_identificador')['item_peso']\
                            .transform(lambda x: x.fillna(x.median()))

print("Nulos restantes após imputação por grupo:", dados['item_peso'].isna().sum())
```

```python
# Rode ANTES de aplicar o fillna acima, ou guarde os identificadores antes
itens_com_peso_global = dados.loc[dados['item_peso'].isna(), 'item_identificador'].unique()
print("Itens que receberam mediana global:", itens_com_peso_global)
```

```python
mediana_global = dados['item_peso'].median()
dados['item_peso'] = dados['item_peso'].fillna(mediana_global)

print(f"Mediana global aplicada: {mediana_global:.3f}")
print("Nulos restantes em item_peso:", dados['item_peso'].isna().sum())
```

```python
print(pd.crosstab(dados['loja_tipo'], dados['loja_tamanho'].isna(), margins=True))

print(pd.crosstab(dados['loja_tipo_localizacao'], dados['loja_tamanho'].isna(), margins=True))

lojas_com_nulo = dados.groupby('loja_identificador')['loja_tamanho'].apply(lambda x: x.isna().all())
print("Lojas 100% sem tamanho reportado:", lojas_com_nulo.sum())
print(lojas_com_nulo[lojas_com_nulo].index.tolist())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

PALETA = "mako"
COR_DESTAQUE = "#2b5c8f"
COR_MEDIANA = "#d95f02"


def formatar_reais_k(valor, pos):
    """Formata valores grandes como R$ XXk para melhor leitura no eixo."""
    if valor >= 1000:
        return f'R$ {valor/1000:.0f}k'
    return f'R$ {valor:.0f}'

def limite_y_percentil(serie, percentil=97):
    """
    Retorna o limite superior do eixo Y com base em um percentil,
    dando margem de 10% para não cortar a caixa do boxplot.
    """
    limite = np.percentile(serie.dropna(), percentil)
    return limite * 1.10


def anotar_medianas(ax, dados, x_col, y_col, ordem, y_offset_frac=0.03):
    """
    Adiciona o valor da mediana como texto acima de cada boxplot/violino,
    seguindo a ordem das categorias no eixo X.
    """
    medianas = dados.groupby(x_col)[y_col].median().reindex(ordem)
    y_max_visivel = ax.get_ylim()[1]
    offset = y_max_visivel * y_offset_frac

    for i, valor in enumerate(medianas):
        if pd.notna(valor):
            ax.text(
                i, valor + offset,
                f'R$ {valor/1000:.0f}k',
                ha='center', va='bottom',
                fontsize=9, fontweight='bold',
                color=COR_MEDIANA
            )


def grafico_vendas_por_tamanho(dados, salvar_como=None):
    ordem = ['Pequeno', 'Médio', 'Grande', 'Desconhecido']

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    sns.boxplot(
        data=dados, x='loja_tamanho', y='vendas_totais',
        order=ordem, palette=PALETA, showfliers=False, ax=axes[0]
    )
    limite_y = limite_y_percentil(dados['vendas_totais'], percentil=97)
    axes[0].set_ylim(0, limite_y)
    anotar_medianas(axes[0], dados, 'loja_tamanho', 'vendas_totais', ordem)

    axes[0].set_title('Distribuição de Vendas (foco na massa de dados)', fontsize=11, fontweight='bold')
    axes[0].set_xlabel('Tamanho da Loja')
    axes[0].set_ylabel('Vendas Totais (R$)')
    axes[0].yaxis.set_major_formatter(ticker.FuncFormatter(formatar_reais_k))

    sns.boxenplot(
        data=dados, x='loja_tamanho', y='vendas_totais',
        order=ordem, palette=PALETA, ax=axes[1]
    )
    axes[1].set_title('Densidade da Distribuição Completa (Boxenplot)', fontsize=11, fontweight='bold')
    axes[1].set_xlabel('Tamanho da Loja')
    axes[1].set_ylabel('')
    axes[1].yaxis.set_major_formatter(ticker.FuncFormatter(formatar_reais_k))

    for ax in axes:
        sns.despine(ax=ax, top=True, right=True)

    fig.suptitle(
        'Vendas Totais por Tamanho de Loja',
        fontsize=15, fontweight='bold', y=1.02
    )
    fig.text(
        0.5, 0.96,
        'Outliers suprimidos no painel A (foco no percentil 97) • Rótulos indicam a mediana por grupo',
        ha='center', fontsize=9, color='gray'
    )

    plt.tight_layout()
    if salvar_como:
        plt.savefig(salvar_como, dpi=150, bbox_inches='tight')
    plt.show()


def grafico_vendas_por_categoria(dados, top_n=None, salvar_como=None):
    resumo = (
        dados.groupby('item_tipo')['vendas_totais']
        .median()
        .sort_values(ascending=True)
    )
    if top_n:
        resumo = resumo.tail(top_n)

    fig, ax = plt.subplots(figsize=(9, max(6, len(resumo) * 0.4)))

    cores = sns.color_palette(PALETA, n_colors=len(resumo))
    barras = ax.barh(resumo.index, resumo.values, color=cores)

    for barra, valor in zip(barras, resumo.values):
        ax.text(
            valor + (resumo.max() * 0.01),
            barra.get_y() + barra.get_height() / 2,
            f'R$ {valor/1000:.0f}k',
            va='center', fontsize=8.5, color='#333333'
        )

    ax.set_title('Mediana de Vendas por Categoria de Produto', fontsize=14, fontweight='bold')
    ax.text(
        0, 1.02,
        'Ranking baseado na mediana (robusto a outliers) • Ordenado do menor para o maior',
        transform=ax.transAxes, fontsize=9, color='gray'
    )
    ax.set_xlabel('Vendas Totais — Mediana (R$)')
    ax.set_ylabel('')
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(formatar_reais_k))
    ax.grid(axis='x', alpha=0.3)
    ax.grid(axis='y', visible=False)

    sns.despine(top=True, right=True)
    plt.tight_layout()
    if salvar_como:
        plt.savefig(salvar_como, dpi=150, bbox_inches='tight')
    plt.show()

def grafico_vendas_por_gordura(dados, salvar_como=None):
    ordem = ['Baixo Teor de Gordura', 'Regular']

    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    sns.violinplot(
        data=dados, x='item_conteudo_gordura', y='vendas_totais',
        order=ordem, palette=PALETA, inner='quartile', ax=axes[0]
    )
    limite_y = limite_y_percentil(dados['vendas_totais'], percentil=97)
    axes[0].set_ylim(0, limite_y)
    axes[0].set_title('Formato da Distribuição (Violin Plot)', fontsize=11, fontweight='bold')
    axes[0].set_xlabel('Teor de Gordura')
    axes[0].set_ylabel('Vendas Totais (R$)')
    axes[0].yaxis.set_major_formatter(ticker.FuncFormatter(formatar_reais_k))

    sns.boxplot(
        data=dados, x='item_conteudo_gordura', y='vendas_totais',
        order=ordem, palette=PALETA, showfliers=False, ax=axes[1]
    )
    axes[1].set_ylim(0, limite_y)
    anotar_medianas(axes[1], dados, 'item_conteudo_gordura', 'vendas_totais', ordem)
    axes[1].set_title('Comparação Direta (Boxplot sem outliers)', fontsize=11, fontweight='bold')
    axes[1].set_xlabel('Teor de Gordura')
    axes[1].set_ylabel('')
    axes[1].yaxis.set_major_formatter(ticker.FuncFormatter(formatar_reais_k))

    for ax in axes:
        sns.despine(ax=ax, top=True, right=True)

    fig.suptitle('Vendas Totais por Teor de Gordura do Produto', fontsize=15, fontweight='bold', y=1.02)
    fig.text(
        0.5, 0.96,
        'Distribuições muito semelhantes indicam que o teor de gordura, isoladamente, tem baixo poder explicativo',
        ha='center', fontsize=9, color='gray'
    )

    plt.tight_layout()
    if salvar_como:
        plt.savefig(salvar_como, dpi=150, bbox_inches='tight')
    plt.show()

# grafico_vendas_por_tamanho(dados)
# grafico_vendas_por_categoria(dados)
# grafico_vendas_por_categoria(dados, top_n=8)
# grafico_vendas_por_gordura(dados)
```
